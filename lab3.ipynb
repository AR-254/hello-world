```python
# Statistical summary
df.describe(include='all')
```

```python

plt.figure(figsize=(6,4))
plt.hist(df['Quantity'].dropna(), bins=30)

plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.title("Distribution of Quantity")

plt.show()
```

```python
# box plot for finding the outliers
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Quantity'])

plt.title("Box Plot of Quantity")
plt.show()
```

```python
top_products = df['Quantity'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_products.plot(kind='bar')

plt.xlabel("Quantity")
plt.ylabel("Count")
plt.title("Top Quantity Distribution")

plt.show()
```

```python
#scatter plot 
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M').dt.hour * 60 + \
             pd.to_datetime(df['Time'], format='%H:%M').dt.minute

# Scatter plot
plt.figure(figsize=(8,5))
plt.scatter(df['Time'], df['Sales'])
plt.xlabel("Time (Minutes from Midnight)")
plt.ylabel("Sales")
plt.title("Scatter Plot: Time vs Sales")
plt.grid()
plt.show()
```

```python

# Heatmap of correlations
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Sort by Date
df = df.sort_values('Date')

# Group sales by date
time_line = df.groupby('Date')['Sales'].sum()

# Line plot (Time Line)
plt.figure(figsize=(10,5))
plt.plot(time_line.index, time_line.values, marker='o')
plt.title("Sales Time Line Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid()
plt.show()
```
